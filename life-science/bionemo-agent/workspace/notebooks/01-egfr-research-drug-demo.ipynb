# Explore an EGFR and gefitinib research question

Review public evidence, explore gefitinib-like candidates, and model one candidate with a public EGFR target. This bounded study uses the bundled public EGFR kinase-domain sequence and gefitinib starting molecule.

> This is an unexecuted launch template, not a saved scientific result. Generated molecules and structures are computational hypotheses for nonclinical research; they do not establish binding, safety, efficacy, therapeutic value, or clinical validity.

## Launch in the BioNeMo agent

Use the workbench's **Open in agent** action, or paste this exact prompt into a new chat:

```text
Please run a small EGFR research study using the bundled public EGFR kinase-domain sequence from UniProt P00533 and gefitinib as the starting molecule. Begin with current public evidence about EGFR and gefitinib resistance, then predict the target structure, generate two gefitinib-like candidates optimized for QED, and model the strongest usable candidate with the same target. Summarize each stage, cite the public sources you used, compare candidate scores and confidence values, link every generated structure, and explain the limitations. This is nonclinical, noncommercial research. I accept the applicable research-use and acceptable-use terms and understand that these computational results do not establish binding, safety, efficacy, therapeutic value, or clinical utility and require independent validation.
```

The public-source review is optional. If it is unavailable, the scientific study can still proceed and should say that the evidence step was skipped. Search snippets are evidence leads, not primary validation.

## Fixed study design

- **Target:** bundled 312-amino-acid EGFR kinase domain from UniProt P00533
- **Starting molecule:** public gefitinib record from PubChem
- **Evidence question:** EGFR/gefitinib resistance mechanisms and medicinal chemistry
- **Candidate search:** two candidates, QED maximization, three iterations, eight particles, minimum similarity 0.4, radius 1
- **Selection:** highest score among candidates with a valid 3D representation; first returned candidate breaks a tie
- **Final model:** selected ligand in complex with the same EGFR sequence

## What to verify

1. Any public evidence is cited and kept separate from inference.
2. Both generated candidates and their optimization scores are reported.
3. The selected candidate follows the stated validity, ranking, and tie-break rules.
4. Scalar confidence values and clickable structure links are included.
5. The conclusion makes no binding, safety, efficacy, therapeutic, or clinical claim.
