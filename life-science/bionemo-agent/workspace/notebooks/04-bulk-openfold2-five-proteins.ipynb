# Predict structures for five public proteins

Fold five familiar public protein sequences and collect a clear result for every record. The bundled FASTA is processed in source order.

> This is an unexecuted launch template with no credentials or saved outputs. The workflow is deliberately sequential and preserves per-record status so one failure does not erase successful results. Predictions require expert review and experimental validation.

## Fixed input fixture

| Record | Protein | Residues | Public source |
|---|---|---:|---|
| `1CRN` | crambin | 46 | RCSB PDB 1CRN |
| `1UBQ` | ubiquitin | 76 | RCSB PDB 1UBQ |
| `1PGA` | protein G | 56 | RCSB PDB 1PGA |
| `2CI2` | chymotrypsin inhibitor 2 | 83 | RCSB PDB 2CI2 |
| `1VII` | villin | 36 | RCSB PDB 1VII |

## Launch in the BioNeMo agent

```text
Please predict structures with OpenFold2 for the five reviewed public sequences in the bundled FASTA: 1CRN, 1UBQ, 1PGA, 2CI2, and 1VII, in that order. Report every protein even if another item fails. For each one, include its status, elapsed time or job reference when available, confidence summary, and a clickable structure link for each successful prediction. Finish with completed and failed counts and explain the scientific limitations. This is nonclinical, noncommercial research. I accept the applicable research-use and acceptable-use terms and understand that these computational predictions make no safety or therapeutic claim and require independent validation.
```

## Reporting expectations

Each of the five records should remain visible in the final summary, including records that could not be completed. Successful records should include their scalar confidence values and a clickable structure link. A failure for one protein should not hide successful results for the others.

## What to verify

1. The bundled input contains exactly five unique IDs, shown above in source order.
2. The result retains one status row per input, including partial failures and remote job identifiers where available.
3. Every successful structure has a clickable 3D link.
4. Completed and failed counts agree with the five per-record results.
5. The summary separates model confidence from experimental validation.
