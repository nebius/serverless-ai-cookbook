# Explore gefitinib-like candidates for EGFR

Generate two gefitinib-like candidates and model the stronger usable candidate with a public EGFR target.

> This is an unexecuted launch template with no credentials or saved outputs. Molecular scores and predicted complexes are computational hypotheses, not evidence of binding, safety, or therapeutic efficacy.

## Launch in the BioNeMo agent

```text
Please explore two gefitinib-like candidates for the bundled public EGFR kinase-domain sequence from UniProt P00533. Optimize the candidates for QED while retaining similarity to gefitinib, select the highest-scoring candidate that has a usable 3D representation, and predict it in complex with the same EGFR target. Compare both candidates, explain why one was selected, report the optimization and confidence values, link every generated structure, and describe the limitations. This is nonclinical, noncommercial research. I accept the applicable research-use and acceptable-use terms and understand that these computational results do not establish binding, safety, efficacy, therapeutic value, or clinical utility and require independent validation.
```

## Fixed study design

- **Target:** bundled 312-amino-acid EGFR kinase domain from UniProt P00533
- **Starting molecule:** `COc1cc2ncnc(Nc3ccc(F)c(Cl)c3)c2cc1OCCCN1CCOCC1`
- **Candidate search:** two candidates, QED maximization, three iterations, eight particles, minimum similarity 0.4, radius 1
- **Selection:** highest score among candidates with a valid 3D representation; first returned candidate breaks a tie
- **Final model:** selected ligand in complex with the same EGFR sequence
